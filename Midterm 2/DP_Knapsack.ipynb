```python
def knapsack(capacity, values, weights):
    n = len(values)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        v = values[i - 1]
        w = weights[i - 1]
        for c in range(capacity + 1):
            if w > c: #can't include
                dp[i][c] = dp[i - 1][c]
            else:
                dp[i][c] = max(dp[i - 1][c], dp[i - 1][c - w] + v)

    return dp[n][capacity], dp

values  = [2, 2, 4, 5, 3]
weights = [3, 1, 3, 4, 2]
capacity = 7

max_value, dp_table = knapsack(capacity, values, weights)
print("Maximum value:", max_value)
print("DP Table:")
for row in dp_table:
    print(row)
```

```python
included_items = []
n = len(values)
c = capacity

while n > 0 and c > 0:
    if dp_table[n][c] != dp_table[n-1][c]:
        included_items.append(n-1)
        c -= weights[n-1]
    n -= 1

included_items.reverse()
print("Items included", included_items)
print("Corresponding values:", [values[idx] for idx in included_items])
print("Corresponding weights:", [weights[idx] for idx in included_items])
```

```python
#Easy

def best_profit(prices):
    smallest_price = prices[0]
    biggest_profit = 0

    for price in prices:
        if price < smallest_price:
            smallest_price = price
        elif price - smallest_price > biggest_profit:
            biggest_profit = price - smallest_price

    return biggest_profit

print(best_profit([7,1,5,3,6,4]))
print(best_profit([7,6,4,3,1]))
```

```python
#Medium

def max_profit_many(prices):
    if len(prices) == 1:
        return 0

    total = 0
    for day in range(1, len(prices)):
        today = prices[day]
        yesterday = prices[day - 1]
        if today > yesterday:
            profit = today - yesterday
            total += profit
    return total

print(max_profit_many([7,1,5,3,6,4]))
print(max_profit_many([1,2,3,4,5]))
print(max_profit_many([7,6,4,3,1]))
```

```python
#Hard

def max_profit_2_trades(prices):
    #buying with the 1st day's price
    first_buy_price = prices[0]
    first_profit = 0
    second_buy_price = prices[0]
    second_profit = 0

    for price in prices:
        first_buy_price = min(first_buy_price, price)           #as cheap as possible
        first_profit = max(first_profit, price - first_buy_price)

        second_buy_price = min(second_buy_price, price - first_profit) #as cheap as possible after 1st profit
        second_profit = max(second_profit, price - second_buy_price)

    return second_profit

print(max_profit_2_trades([3,3,5,0,0,3,1,4]))
print(max_profit_2_trades([1,2,3,4,5]))
print(max_profit_2_trades([7,6,4,3,1]))
```

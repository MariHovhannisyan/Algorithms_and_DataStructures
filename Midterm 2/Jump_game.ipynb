```python
def min_jumps(nums):
    n = len(nums)
    if n == 1:     #at the end
        return 0

    jumps = 0
    end_of_jump = 0
    farthest = 0

    for i in range(n - 1):
        farthest = max(farthest, i + nums[i])
        if i == end_of_jump:
            jumps += 1
            end_of_jump = farthest
            if end_of_jump >= n - 1: #we can reach or pass the end, we can stop
                break

    return jumps

# Example usage:
print(min_jumps([2,3,1,1,4]))
print(min_jumps([2,3,0,1,4]))
print(min_jumps([1,2,1,1,1]))
```

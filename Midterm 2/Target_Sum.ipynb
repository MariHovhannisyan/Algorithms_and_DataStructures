```python
def target_sum(nums, target):
        total_sum = sum(nums)
        if abs(target) > total_sum or (total_sum + target) % 2 == 1:
            return 0

        s1 = (total_sum + target) // 2

        dp = [0] * (s1 + 1)
        dp[0] = 1

        for num in nums:
            for j in range(s1, num - 1, -1):
                dp[j] += dp[j - num]

        return dp[s1]

print(target_sum([1,1,1,1,1], 3))
print(target_sum([1,2,3,4,5], -3))
print(target_sum([1], 1))
```

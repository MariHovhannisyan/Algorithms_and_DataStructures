```python
def find_extra_edge(connections):
    n = len(connections)
    parent = [i for i in range(n + 1)]

    def find_parent(node):
        while parent[node] != node:
            parent[node] = parent[parent[node]]
            node = parent[node]
        return node

    for edge in connections:
        node1, node2 = edge
        root1 = find_parent(node1)
        root2 = find_parent(node2)
        if root1 == root2:  #found cycle
            return edge
        else:
            parent[root2] = root1

edges1 = [[1,2],[1,3],[2,3]]
print(find_extra_edge(edges1))

edges2 = [[1,2],[2,3],[3,4],[1,4],[1,5]]
print(find_extra_edge(edges2))
```

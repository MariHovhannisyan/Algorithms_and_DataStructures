```python
def path_exists(n, edges, start, end):
    #each node has a list of neighbors
    neighbors = {}
    for i in range(n):
        neighbors[i] = []
    for u, v in edges:
        neighbors[u].append(v)
        neighbors[v].append(u)

    visited = set()
    to_visit = [start]
    visited.add(start)

    while to_visit:
        current = to_visit.pop(0)
        if current == end:
            return True  #found a way
        for next_node in neighbors[current]:
            if next_node not in visited:
                visited.add(next_node)
                to_visit.append(next_node)
    return False

n1 = 3
edges1 = [[0,1],[1,2],[2,0]]
print(path_exists(n1, edges1, 0, 2))

n2 = 6
edges2 = [[0,1],[0,2],[3,5],[5,4],[4,3]]
print(path_exists(n2, edges2, 0, 5))
```

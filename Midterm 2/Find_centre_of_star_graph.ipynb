```python
#in a star graph the centre node is the only node that appears in every edge
def find_centre(edges):
    if edges[0][0] in edges[1]:
        return edges[0][0]
    else:
        return edges[0][1]

edges1 = [[1,2],[2,3],[4,2]]
print(find_centre(edges1))

edges2 = [[1,2],[5,1],[1,3],[1,4]]
print(find_centre(edges2))
```
